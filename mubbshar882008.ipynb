```python
class books001:
    def __init__(self,book,writer,author,duedate,return_date):
        self.book=book, #publics
        self.writer=writer, #public
        self.author=author #public
        self.duedate=duedate #pyblic
        self.return_date=return_date
```

```python
book=books001("death_and_soul","albert_eienstien","john",'15-12-13','20-12-22')
```

```python

import datetime
# Enter due date and return date
due_date = datetime.datetime.strptime(book.duedate, "%y-%m-%d").date()
return_date = datetime.datetime.strptime(book.return_date,"%y-%m-%d").date()
# Fine per day
fine_per_day = 10

# Calculate late days
late_days = (return_date - due_date).days

# Calculate fine
if late_days>0:
    fine1 = late_days*fine_per_day
else:
    fine1 = 0

print("Late days:", max(late_days, 0))
print("Total fine: Rs.", fine1)
fine1=fine1
```
